## Original model: fundamentals-only MLP, single 70/15/15 split

```python
# ==========================================
# FUNDAMENTALS-ONLY BASELINE
# ==========================================

feature_cols = [
    "Revenue",
    "Revenue Growth (YoY)",
    "Net Income",
    "Net Income Growth",
    "EPS (Basic)",
    "EPS (Diluted)",
    "EPS Growth",
    "Total Assets",
    "Cash & Equivalents"
]

target_col = "CLOSE"

X = aligned_data[feature_cols].copy()
y = aligned_data[target_col].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
import random

# ========================================================
# REPRODUCIBILITY
# ========================================================

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)
```

```python
# ==========================================
# 1. NEXT-DAY TARGET (FUNDAMENTALS-ONLY INPUTS)
# ==========================================
df = aligned_data.sort_values("DATE").reset_index(drop=True).copy()

# Target = next trading day's close
df["next_close"] = df["CLOSE"].shift(-1)

# Date being predicted
df["tdate"] = df["DATE"].shift(-1)

# Next-day log return
df["target_ret"] = np.log(
    df["next_close"] / df["CLOSE"]
)

# Remove rows with missing target/features
df = df.dropna(
    subset=["next_close"] + feature_cols
).reset_index(drop=True)
```

```python
# ==========================================
# 2. INPUTS AND TARGETS
# ==========================================

X = df[feature_cols]

# Next-day closing price
y_price = df["next_close"]

# Next-day log return
y_return = df["target_ret"]

# Today's close
# Used only for converting predicted return → price
# NOT used as an MLP input
current_close = df["CLOSE"]

target_dates = pd.to_datetime(df["tdate"])
```

```python
# ==========================================
# 3. CHRONOLOGICAL TRAIN / VAL / TEST SPLIT
# ==========================================

n = len(df)

train_end = int(n * 0.70)
val_end = int(n * 0.85)

X_train = X.iloc[:train_end]
X_val   = X.iloc[train_end:val_end]
X_test  = X.iloc[val_end:]

y_price_train = y_price.iloc[:train_end]
y_price_val   = y_price.iloc[train_end:val_end]
y_price_test  = y_price.iloc[val_end:]

y_return_train = y_return.iloc[:train_end]
y_return_val   = y_return.iloc[train_end:val_end]
y_return_test  = y_return.iloc[val_end:]

current_close_test = current_close.iloc[val_end:]

print("\nDataset Split:")
print("Train:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)

print("\nDate Ranges:")
print(
    "Train:",
    df["DATE"].iloc[0],
    "to",
    df["DATE"].iloc[train_end - 1]
)

print(
    "Validation:",
    df["DATE"].iloc[train_end],
    "to",
    df["DATE"].iloc[val_end - 1]
)

print(
    "Test:",
    df["DATE"].iloc[val_end],
    "to",
    df["DATE"].iloc[-1]
)
```

```python
from sklearn.preprocessing import StandardScaler

# ==========================================
# 4. FEATURE SCALING
# ==========================================

scaler_X = StandardScaler()

X_train_scaled = scaler_X.fit_transform(X_train)
X_val_scaled   = scaler_X.transform(X_val)
X_test_scaled  = scaler_X.transform(X_test)

# Scale target using training data only
scaler_y_price = StandardScaler()

y_price_train_scaled = scaler_y_price.fit_transform(
    y_price_train.values.reshape(-1, 1)
)

y_price_val_scaled = scaler_y_price.transform(
    y_price_val.values.reshape(-1, 1)
)

# Scale target using training data only
scaler_y_return = StandardScaler()

y_return_train_scaled = scaler_y_return.fit_transform(
    y_return_train.values.reshape(-1, 1)
)

y_return_val_scaled = scaler_y_return.transform(
    y_return_val.values.reshape(-1, 1)
)
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import Sequential, layers, callbacks
from sklearn.metrics import (
    mean_absolute_error as MAE,
    mean_squared_error as MSE,
    r2_score
)

# ==========================================
# 5. METRICS
# ==========================================

def report(name, actual, predicted, previous_close):

    mae = MAE(actual, predicted)
    rmse = np.sqrt(MSE(actual, predicted))

    mape = np.mean(
        np.abs((actual - predicted) / actual)
    ) * 100

    r2 = r2_score(actual, predicted)

    directional_accuracy = np.mean(
        np.sign(actual - previous_close)
        ==
        np.sign(predicted - previous_close)
    ) * 100

    print(
        f"{name}: "
        f"MAE {mae:.2f} "
        f"RMSE {rmse:.2f} "
        f"MAPE {mape:.2f}% "
        f"R2 {r2:.3f} "
        f"DirAcc {directional_accuracy:.1f}%"
    )


# ==========================================
# 6. MLP MODEL
# ==========================================

def build_mlp():

    tf.random.set_seed(SEED)
    np.random.seed(SEED)

    model = Sequential([
        layers.Input(shape=(X_train_scaled.shape[1],)),

        layers.Dense(64, activation="relu"),

        layers.Dropout(0.2),

        layers.Dense(32, activation="relu"),

        layers.Dense(1)
    ])

    model.compile(
        optimizer="adam",
        loss="mse"
    )

    return model


# ==========================================
# 7. MODEL A — NEXT-DAY CLOSE
# ==========================================

model_price = build_mlp()

history_price = model_price.fit(
    X_train_scaled,
    y_price_train_scaled,
    validation_data=(
        X_val_scaled,
        y_price_val_scaled
    ),
    epochs=200,
    batch_size=32,
    verbose=0,
    callbacks=[
        callbacks.EarlyStopping(
            patience=20,
            restore_best_weights=True
        )
    ]
)

# ==========================================
# 8. MODEL B — NEXT-DAY LOG RETURN
# ==========================================
model_return = build_mlp()

history_return = model_return.fit(
    X_train_scaled,
    y_return_train_scaled,
    validation_data=(
        X_val_scaled,
        y_return_val_scaled
    ),
    epochs=200,
    batch_size=32,
    verbose=0,
    callbacks=[
        callbacks.EarlyStopping(
            patience=20,
            restore_best_weights=True
        )
    ]
)

# ==========================================
# 9. PREDICTIONS — TRAIN / VALIDATION / TEST
# ==========================================

# ---------- MODEL A: PRICE TARGET ----------

pred_price_train_scaled = model_price.predict(
    X_train_scaled,
    verbose=0
)

pred_price_val_scaled = model_price.predict(
    X_val_scaled,
    verbose=0
)

pred_price_test_scaled = model_price.predict(
    X_test_scaled,
    verbose=0
)

# Convert back to actual prices
pred_price_train = scaler_y_price.inverse_transform(
    pred_price_train_scaled
).ravel()

pred_price_val = scaler_y_price.inverse_transform(
    pred_price_val_scaled
).ravel()

pred_price_test = scaler_y_price.inverse_transform(
    pred_price_test_scaled
).ravel()


# ---------- MODEL B: LOG-RETURN TARGET ----------

pred_return_train_scaled = model_return.predict(
    X_train_scaled,
    verbose=0
)

pred_return_val_scaled = model_return.predict(
    X_val_scaled,
    verbose=0
)

pred_return_test_scaled = model_return.predict(
    X_test_scaled,
    verbose=0
)

# Convert back to actual log returns
pred_return_train = scaler_y_return.inverse_transform(
    pred_return_train_scaled
).ravel()

pred_return_val = scaler_y_return.inverse_transform(
    pred_return_val_scaled
).ravel()

pred_return_test = scaler_y_return.inverse_transform(
    pred_return_test_scaled
).ravel()


# Convert predicted log return → predicted next-day price
pred_return_price_train = (
    current_close.iloc[:train_end].values
    * np.exp(pred_return_train)
)

pred_return_price_val = (
    current_close.iloc[train_end:val_end].values
    * np.exp(pred_return_val)
)

pred_return_price_test = (
    current_close_test.values
    * np.exp(pred_return_test)
)


# ==========================================
# 10. TRAIN / VALIDATION / TEST METRICS
# ==========================================

print("\nFUNDAMENTALS-ONLY MLP — ALL SPLITS")
print("=" * 70)

# ---------- MODEL A ----------
print("\nMODEL A — NEXT-DAY CLOSE (PRICE TARGET)")
print("-" * 70)

report(
    "Train",
    y_price_train.values,
    pred_price_train,
    current_close.iloc[:train_end].values
)

report(
    "Validation",
    y_price_val.values,
    pred_price_val,
    current_close.iloc[train_end:val_end].values
)

report(
    "Test",
    y_price_test.values,
    pred_price_test,
    current_close_test.values
)


# ---------- MODEL B ----------
print("\nMODEL B — NEXT-DAY LOG RETURN")
print("-" * 70)

report(
    "Train",
    y_price_train.values,
    pred_return_price_train,
    current_close.iloc[:train_end].values
)

report(
    "Validation",
    y_price_val.values,
    pred_return_price_val,
    current_close.iloc[train_end:val_end].values
)

report(
    "Test",
    y_price_test.values,
    pred_return_price_test,
    current_close_test.values
)


# ==========================================
# 10. SAVE TEST PREDICTIONS
# ==========================================

results = pd.DataFrame({
    "DATE": target_dates.iloc[val_end:].values,
    "Actual Close": y_price_test.values,
    "MLP A (Price Target)": pred_price_test,
    "MLP B (Return Target)": pred_return_price_test
})

print("\nTest Predictions:")
print(results.head(20))

results.to_csv(
    "fundamentals_mlp_nextday_predictions.csv",
    index=False
)


# ==========================================
# 11. PLOT
# ==========================================

plt.figure(figsize=(12, 5))

plt.plot(
    results["DATE"],
    results["Actual Close"],
    label="Actual"
)

plt.plot(
    results["DATE"],
    results["MLP A (Price Target)"],
    label="MLP A (Price)"
)

plt.plot(
    results["DATE"],
    results["MLP B (Return Target)"],
    label="MLP B (Return)"
)

plt.legend()
plt.title(
    "Fundamentals-only MLP: Next-day Close — Test Period"
)
plt.tight_layout()
plt.show()
```

## B. Walk-forward validation

## Walk-Forward Validation (new section)

Same data, same features, same MLP architecture/optimizer/early-stopping as above.
Only the **evaluation scheme** changes: instead of one fixed 70/15/15 split, the model is
re-trained repeatedly on past data and tested on the *next* block of days, then the window moves forward.

```python
# ==========================================
# WF-1. CONFIG
# ==========================================
WF_START        = val_end   # first fold's test block starts where the old single test set started (same test days -> comparable)
WF_STEP         = 21        # test block per fold ~ 1 trading month; model is re-trained every 21 days
WF_INNER_VAL    = 0.15      # last 15% of each training window is held out (chronologically) for early stopping
WF_ROLL_WINDOW  = val_end   # rolling mode: train on the most recent 998 rows only (= size of fold-1 training window)

print("Total rows:", len(df))
print("First test row:", df['DATE'].iloc[WF_START].date(), "| Last:", df['DATE'].iloc[-1].date())
print("Number of folds:", int(np.ceil((len(df) - WF_START) / WF_STEP)))
```

```python
# ==========================================
# WF-2. SAME MLP (only change: n_features is an argument, not a global)
# ==========================================
def build_mlp_wf(n_features):
    tf.random.set_seed(SEED)
    np.random.seed(SEED)
    model = Sequential([
        layers.Input(shape=(n_features,)),
        layers.Dense(64, activation="relu"),
        layers.Dropout(0.2),
        layers.Dense(32, activation="relu"),
        layers.Dense(1)
    ])
    model.compile(optimizer="adam", loss="mse")
    return model


def fit_predict_fold(tr_idx, te_idx):
    """Train Model A (price) and Model B (log-return) on tr_idx, predict te_idx.
       Everything (scalers, early-stopping val set) is built from tr_idx only."""
    n_tr = len(tr_idx)
    cut  = int(n_tr * (1 - WF_INNER_VAL))
    fit_idx, val_idx = tr_idx[:cut], tr_idx[cut:]

    # --- scale features: fit on fit_idx ONLY ---
    sx = StandardScaler()
    Xf = sx.fit_transform(X.iloc[fit_idx])
    Xv = sx.transform(X.iloc[val_idx])
    Xt = sx.transform(X.iloc[te_idx])

    out = {}
    for tag, y_series in [("A", y_price), ("B", y_return)]:
        sy = StandardScaler()
        yf = sy.fit_transform(y_series.iloc[fit_idx].values.reshape(-1, 1))
        yv = sy.transform(y_series.iloc[val_idx].values.reshape(-1, 1))

        m = build_mlp_wf(Xf.shape[1])
        m.fit(
            Xf, yf,
            validation_data=(Xv, yv),
            epochs=200, batch_size=32, verbose=0,
            callbacks=[callbacks.EarlyStopping(patience=20, restore_best_weights=True)]
        )
        p = sy.inverse_transform(m.predict(Xt, verbose=0)).ravel()
        out[tag] = p

    cur = current_close.iloc[te_idx].values
    return out["A"], cur * np.exp(out["B"]), cur
```

```python
# ==========================================
# WF-3. WALK-FORWARD LOOP
# ==========================================
def walk_forward(mode="expanding"):
    rows = []
    fold = 0
    for start in range(WF_START, len(df), WF_STEP):
        end    = min(start + WF_STEP, len(df))
        tr_lo  = 0 if mode == "expanding" else max(0, start - WF_ROLL_WINDOW)
        tr_idx = np.arange(tr_lo, start)       # strictly BEFORE the test block
        te_idx = np.arange(start, end)

        pA, pB, cur = fit_predict_fold(tr_idx, te_idx)

        fold += 1
        rows.append(pd.DataFrame({
            "fold":        fold,
            "DATE":        target_dates.iloc[te_idx].values,
            "prev_close":  cur,
            "actual":      y_price.iloc[te_idx].values,
            "pred_A":      pA,
            "pred_B":      pB,
            "train_size":  len(tr_idx),
            "train_start": df["DATE"].iloc[tr_lo],
            "train_end":   df["DATE"].iloc[start - 1],
        }))
        print(f"[{mode}] fold {fold:>2} | train {len(tr_idx):>4} rows "
              f"({df['DATE'].iloc[tr_lo].date()} -> {df['DATE'].iloc[start-1].date()}) "
              f"| test {len(te_idx):>2} rows ({df['DATE'].iloc[start].date()} -> {df['DATE'].iloc[end-1].date()})")
    return pd.concat(rows, ignore_index=True)

wf_expanding = walk_forward("expanding")
wf_rolling   = walk_forward("rolling")
```

```python
# ==========================================
# WF-4. METRICS (same definitions as report(), plus a naive baseline)
# ==========================================
def metrics(actual, pred, prev):
    actual, pred, prev = map(np.asarray, (actual, pred, prev))
    mae  = MAE(actual, pred)
    rmse = np.sqrt(MSE(actual, pred))
    mape = np.mean(np.abs((actual - pred) / actual)) * 100
    r2   = r2_score(actual, pred)
    # direction is undefined for "no change" predictions (naive baseline) -> NaN
    pd_  = np.sign(pred - prev)
    dacc = np.mean(np.sign(actual - prev) == pd_) * 100 if np.any(pd_ != 0) else np.nan
    return dict(MAE=mae, RMSE=rmse, MAPE=mape, R2=r2, DirAcc=dacc)

def summarize(wf, label):
    rows = {
        f"{label} | Model A (price)":  metrics(wf.actual, wf.pred_A, wf.prev_close),
        f"{label} | Model B (return)": metrics(wf.actual, wf.pred_B, wf.prev_close),
        f"{label} | Naive (yesterday)": metrics(wf.actual, wf.prev_close, wf.prev_close),
    }
    return pd.DataFrame(rows).T

# single-split result on the SAME test rows (from the cells above)
single = pd.DataFrame({
    "Single split | Model A (price)":  metrics(y_price_test.values, pred_price_test, current_close_test.values),
    "Single split | Model B (return)": metrics(y_price_test.values, pred_return_price_test, current_close_test.values),
}).T

summary = pd.concat([
    single,
    summarize(wf_expanding, "WF expanding"),
    summarize(wf_rolling,   "WF rolling"),
])

up_rate = np.mean(wf_expanding.actual > wf_expanding.prev_close) * 100
print(f"Test rows: {len(wf_expanding)} | Days the price actually went UP: {up_rate:.1f}% (a 'always up' guess scores this)")
print()
print(summary.round(3).to_string())
```

```python
# ==========================================
# WF-5. PER-FOLD STABILITY
# ==========================================
def per_fold(wf):
    out = []
    for f, g in wf.groupby("fold"):
        a = metrics(g.actual, g.pred_A, g.prev_close)
        b = metrics(g.actual, g.pred_B, g.prev_close)
        n = metrics(g.actual, g.prev_close, g.prev_close)
        out.append(dict(
            fold=f, test_from=pd.to_datetime(g.DATE.iloc[0]).date(), n=len(g),
            A_MAE=a["MAE"], A_MAPE=a["MAPE"], A_Dir=a["DirAcc"],
            B_MAE=b["MAE"], B_MAPE=b["MAPE"], B_Dir=b["DirAcc"],
            Naive_MAE=n["MAE"]
        ))
    return pd.DataFrame(out)

fold_exp = per_fold(wf_expanding)
fold_rol = per_fold(wf_rolling)

print("EXPANDING WINDOW — per fold"); print(fold_exp.round(2).to_string(index=False))
print("\nROLLING WINDOW — per fold");   print(fold_rol.round(2).to_string(index=False))

print("\nMean ± std across folds (expanding):")
print(fold_exp[["A_MAE","B_MAE","Naive_MAE","A_Dir","B_Dir"]].agg(["mean","std"]).round(2).to_string())
```

```python
# ==========================================
# WF-6. PLOTS
# ==========================================
fig, axes = plt.subplots(2, 1, figsize=(13, 9), sharex=True)

for ax, wf, title in [(axes[0], wf_expanding, "Expanding window"),
                      (axes[1], wf_rolling,   "Rolling window")]:
    d = pd.to_datetime(wf["DATE"])
    ax.plot(d, wf.actual, label="Actual", color="black", lw=1.5)
    ax.plot(d, wf.pred_A, label="Model A (price target)", alpha=.8)
    ax.plot(d, wf.pred_B, label="Model B (return target)", alpha=.8)
    for s in wf.groupby("fold").DATE.first().values[1:]:
        ax.axvline(pd.to_datetime(s), color="grey", ls=":", lw=.8)   # re-train points
    ax.set_title(f"Walk-forward — {title} (dotted = model re-trained)")
    ax.legend()

plt.tight_layout(); plt.show()

# MAE per fold vs naive
fig, ax = plt.subplots(figsize=(11, 4))
x = np.arange(len(fold_exp)); w = .27
ax.bar(x - w, fold_exp.A_MAE,     w, label="Model A")
ax.bar(x,     fold_exp.B_MAE,     w, label="Model B")
ax.bar(x + w, fold_exp.Naive_MAE, w, label="Naive (yesterday's close)")
ax.set_xticks(x); ax.set_xticklabels(fold_exp.fold)
ax.set_xlabel("Fold"); ax.set_ylabel("MAE (PKR)")
ax.set_title("Per-fold MAE — expanding window")
ax.legend(); plt.tight_layout(); plt.show()

wf_expanding.to_csv("walkforward_expanding_predictions.csv", index=False)
wf_rolling.to_csv("walkforward_rolling_predictions.csv", index=False)
```

## C. Multi-horizon forecasting

## Multi-Horizon Forecasting (new section)

Same data, same 9 fundamentals features, same MLP (Model A = price target, Model B = log-return target).
Only change: instead of always predicting **t+1**, a separate copy of the model is trained for each horizon
**h = 1, 5, 10, 21 trading days ahead** ("direct" multi-horizon forecasting).
Train/val/test boundaries are the same rows as the original 70/15/15 split, and a purge gap is applied so
overlapping target windows cannot leak across split boundaries.

```python
# ==========================================
# MH-1. CONFIG + HORIZON-SPECIFIC DATA
# ==========================================
from sklearn.preprocessing import StandardScaler

HORIZONS     = [1, 5, 10, 21]   # trading days ahead: 1 day, ~1 week, ~2 weeks, ~1 month
MH_TRAIN_END = train_end        # same row boundaries as your original split
MH_VAL_END   = val_end

base = aligned_data.sort_values("DATE").reset_index(drop=True).copy()

def build_horizon_frame(h):
    """Same construction as the original next-day target, but shifted by h days."""
    d = base.copy()
    d["target_close"] = d["CLOSE"].shift(-h)                       # close h days later
    d["tdate"]        = d["DATE"].shift(-h)                        # date being predicted
    d["target_ret"]   = np.log(d["target_close"] / d["CLOSE"])     # h-day log return
    return d.dropna(subset=["target_close"] + feature_cols).reset_index(drop=True)

d1 = build_horizon_frame(1)
print("Sanity: h=1 frame has", len(d1), "rows; original df has", len(df), "rows")
for h in HORIZONS:
    print(f"h={h:>2}: usable rows = {len(build_horizon_frame(h))}")
```

```python
# ==========================================
# MH-2. SAME MLP + GENERIC FIT/PREDICT HELPER
# ==========================================
def build_mlp_mh(n_features):
    tf.random.set_seed(SEED)
    np.random.seed(SEED)
    model = Sequential([
        layers.Input(shape=(n_features,)),
        layers.Dense(64, activation="relu"),
        layers.Dropout(0.2),
        layers.Dense(32, activation="relu"),
        layers.Dense(1)
    ])
    model.compile(optimizer="adam", loss="mse")
    return model


def mh_fit_predict(d, fit_idx, val_idx, te_idx):
    """Train Model A and Model B on fit_idx (early stopping on val_idx), predict te_idx.
       Scalers are fit on fit_idx only. Returns predicted PRICE for A and B, plus today's close."""
    sx = StandardScaler()
    Xf = sx.fit_transform(d[feature_cols].iloc[fit_idx])
    Xv = sx.transform(d[feature_cols].iloc[val_idx])
    Xt = sx.transform(d[feature_cols].iloc[te_idx])
    cur = d["CLOSE"].iloc[te_idx].values

    out = {}
    for tag, col in [("A", "target_close"), ("B", "target_ret")]:
        sy = StandardScaler()
        yf = sy.fit_transform(d[col].iloc[fit_idx].values.reshape(-1, 1))
        yv = sy.transform(d[col].iloc[val_idx].values.reshape(-1, 1))

        m = build_mlp_mh(Xf.shape[1])
        m.fit(
            Xf, yf,
            validation_data=(Xv, yv),
            epochs=200, batch_size=32, verbose=0,
            callbacks=[callbacks.EarlyStopping(patience=20, restore_best_weights=True)]
        )
        out[tag] = sy.inverse_transform(m.predict(Xt, verbose=0)).ravel()

    return out["A"], cur * np.exp(out["B"]), cur
```

```python
# ==========================================
# MH-3. TRAIN ONE MODEL PER HORIZON (single chronological split, with purge)
# ==========================================
mh_frames = []

for h in HORIZONS:
    d = build_horizon_frame(h)
    purge = h - 1     # a training row's target window (t, t+h] must not overlap the next split's windows

    fit_idx = np.arange(0, MH_TRAIN_END - purge)
    val_idx = np.arange(MH_TRAIN_END, MH_VAL_END - purge)
    te_idx  = np.arange(MH_VAL_END, len(d))

    pA, pB, cur = mh_fit_predict(d, fit_idx, val_idx, te_idx)

    mh_frames.append(pd.DataFrame({
        "h":          h,
        "DATE":       d["tdate"].iloc[te_idx].values,
        "prev_close": cur,
        "actual":     d["target_close"].iloc[te_idx].values,
        "pred_A":     pA,
        "pred_B":     pB,
    }))
    print(f"h={h:>2} | fit {len(fit_idx)} | val {len(val_idx)} | test {len(te_idx)} rows "
          f"({d['DATE'].iloc[te_idx[0]].date()} -> predicting up to {pd.to_datetime(d['tdate'].iloc[te_idx[-1]]).date()})")

mh_pred = pd.concat(mh_frames, ignore_index=True)
mh_pred.to_csv("multihorizon_predictions.csv", index=False)
```

```python
# ==========================================
# MH-4. METRICS (price-space like report(), plus return-space and skill vs naive)
# ==========================================
def mh_metrics(actual, pred, prev):
    actual, pred, prev = map(np.asarray, (actual, pred, prev))
    mae  = MAE(actual, pred)
    rmse = np.sqrt(MSE(actual, pred))
    mape = np.mean(np.abs((actual - pred) / actual)) * 100
    r2   = r2_score(actual, pred)                       # price-space R2 (inflated by trending prices)

    sgn_pred = np.sign(pred - prev)
    dacc = np.mean(np.sign(actual - prev) == sgn_pred) * 100 if np.any(sgn_pred != 0) else np.nan

    a_ret = np.log(actual / prev)                       # realised h-day log return
    p_ret = np.log(pred / prev)                         # predicted h-day log return
    r2_ret = r2_score(a_ret, p_ret)                     # return-space R2 (the honest one)
    corr   = np.corrcoef(a_ret, p_ret)[0, 1] if np.std(p_ret) > 0 else np.nan
    return dict(MAE=mae, RMSE=rmse, MAPE=mape, R2_price=r2, DirAcc=dacc, R2_return=r2_ret, Corr_return=corr)


def mh_summary(pred_df):
    rows = {}
    for h, g in pred_df.groupby("h"):
        for name, p in [("Model A (price)",  g.pred_A),
                        ("Model B (return)", g.pred_B),
                        ("Naive (no change)", g.prev_close)]:
            r = mh_metrics(g.actual.values, p.values, g.prev_close.values)
            r["n_test"] = len(g)
            rows[(h, name)] = r
    out = pd.DataFrame(rows).T
    out.index.names = ["horizon", "model"]
    naive_mae = out.xs("Naive (no change)", level="model")["MAE"]
    out["MAE_vs_naive"] = [out.loc[(h, m), "MAE"] / naive_mae[h] for h, m in out.index]   # <1 beats naive
    return out

mh_table = mh_summary(mh_pred)
up_rate = mh_pred.assign(up=lambda x: x.actual > x.prev_close).groupby("h").up.mean() * 100

print("Share of test days where price was HIGHER h days later (what an 'always up' guess scores):")
print(up_rate.round(1).to_string())
print()
print(mh_table.round(3).to_string())
```

```python
# ==========================================
# MH-5. PLOTS
# ==========================================
tbl = mh_table.reset_index()
models = ["Model A (price)", "Model B (return)", "Naive (no change)"]

fig, axes = plt.subplots(1, 3, figsize=(17, 4.5))

for m in models:
    t = tbl[tbl.model == m]
    axes[0].plot(t.horizon, t.MAE, marker="o", label=m)
axes[0].set_title("MAE vs horizon"); axes[0].set_xlabel("Horizon (trading days)"); axes[0].set_ylabel("MAE (PKR)")
axes[0].legend()

for m in models[:2]:
    t = tbl[tbl.model == m]
    axes[1].plot(t.horizon, t.MAE_vs_naive, marker="o", label=m)
axes[1].axhline(1.0, color="black", ls="--", lw=1)
axes[1].set_title("MAE / Naive MAE  (below 1 = beats naive)")
axes[1].set_xlabel("Horizon (trading days)"); axes[1].legend()

for m in models[:2]:
    t = tbl[tbl.model == m]
    axes[2].plot(t.horizon, t.DirAcc, marker="o", label=m)
axes[2].plot(up_rate.index, up_rate.values, marker="s", ls="--", color="grey", label="Always-up baseline")
axes[2].axhline(50, color="black", ls=":", lw=1)
axes[2].set_title("Direction accuracy vs horizon (%)")
axes[2].set_xlabel("Horizon (trading days)"); axes[2].legend()

plt.tight_layout(); plt.show()

# predictions vs actual, one panel per horizon
fig, axes = plt.subplots(len(HORIZONS), 1, figsize=(13, 3.2 * len(HORIZONS)), sharex=False)
for ax, h in zip(np.atleast_1d(axes), HORIZONS):
    g = mh_pred[mh_pred.h == h]
    x = pd.to_datetime(g.DATE)
    ax.plot(x, g.actual, color="black", lw=1.5, label="Actual")
    ax.plot(x, g.pred_A, alpha=.8, label="Model A (price)")
    ax.plot(x, g.pred_B, alpha=.8, label="Model B (return)")
    ax.set_title(f"Horizon h = {h} trading day(s)")
    ax.legend(loc="upper left", fontsize=8)
plt.tight_layout(); plt.show()
```

### Optional: multi-horizon **inside** walk-forward (most robust, slowest)

Set `RUN_WF_MH = True` to re-train every `MH_WF_STEP` days for every horizon.
It trains about 4 horizons × 9 folds × 2 models = 72 networks, so it takes a while.
Expanding window only. The purge gap is applied both before each test block and inside each fold's train/val split.

```python
# ==========================================
# MH-6. (OPTIONAL) WALK-FORWARD x MULTI-HORIZON
# ==========================================
RUN_WF_MH       = True   # <- set True to run
MH_WF_STEP      = 21
MH_WF_INNER_VAL = 0.15

if RUN_WF_MH:
    wfmh_frames = []
    for h in HORIZONS:
        d = build_horizon_frame(h)
        purge = h - 1
        for fold, start in enumerate(range(MH_VAL_END, len(d), MH_WF_STEP), start=1):
            end  = min(start + MH_WF_STEP, len(d))
            pool = np.arange(0, start - purge)                 # training pool: targets must end before this test block
            cut  = int(len(pool) * (1 - MH_WF_INNER_VAL))
            fit_idx = pool[:max(cut - purge, 1)]               # purge between fit and inner-val
            val_idx = pool[cut:]
            te_idx  = np.arange(start, end)

            pA, pB, cur = mh_fit_predict(d, fit_idx, val_idx, te_idx)
            wfmh_frames.append(pd.DataFrame({
                "h": h, "fold": fold,
                "DATE": d["tdate"].iloc[te_idx].values,
                "prev_close": cur,
                "actual": d["target_close"].iloc[te_idx].values,
                "pred_A": pA, "pred_B": pB,
            }))
            print(f"h={h:>2} fold {fold:>2} | fit {len(fit_idx)} | val {len(val_idx)} | test {len(te_idx)}")

    wfmh_pred = pd.concat(wfmh_frames, ignore_index=True)
    wfmh_pred.to_csv("walkforward_multihorizon_predictions.csv", index=False)

    wfmh_table = mh_summary(wfmh_pred)
    print("\nWALK-FORWARD x MULTI-HORIZON (expanding window)")
    print(wfmh_table.round(3).to_string())

    # side by side with the single-split multi-horizon result
    cmp = pd.concat({"single split": mh_table, "walk-forward": wfmh_table}, axis=1)
    print("\nMAE vs naive, single split vs walk-forward")
    print(cmp.xs("MAE_vs_naive", axis=1, level=1).round(3).to_string())
```
